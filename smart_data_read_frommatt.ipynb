```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Created on Tue Sep  2 20:25:17 2025

@author: Matt
"""
import rasterio
import numpy as np

file_path = "Great_Salt_Lake_TBDEM.tif"

with rasterio.open(file_path) as src:
    # Downsample by 20
    scale = 20
    data = src.read(
        1,
        out_shape=(src.height // scale, src.width // scale)
    )
    
    # Calculate new transform
    transform = src.transform * src.transform.scale(
        (src.width / data.shape[1]),
        (src.height / data.shape[0])
    )
    
    crs = src.crs  # coordinate system

# Save array + metadata together
np.savez_compressed(
    "great_salt_lake_downsampled_with_meta.npz",
    data=data,
    transform=transform.to_gdal(),  # save affine as tuple
    crs=crs.to_wkt()                # save CRS in text form
)
```
